# Netflix Content Analysis
### Python Data Cleaning + Excel Dashboard

Objective:
Analyze Netflix content trends based on movies, TV shows, genres, countries, and ratings to uncover business insights.

```python
import numpy as np 
import pandas as pd
```

```python
df = pd.read_csv("Netflix.csv")
df
```

```python
df.shape
df.info()
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df["director"] = df["director"].fillna("Unknown")
```

```python
df["cast"] = df["cast"].fillna("Unknown")
```

```python
df["country"] = df["country"].fillna("Unknown")
```

```python
df = df.dropna(subset = ['date_added','rating','duration'])
```

```python
df.isna().sum()
```

```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
df["year_added"] = df["date_added"].dt.year
df["year_added"]=df["year_added"].fillna(0).astype('int32')
```

```python
df["genre"] = df["listed_in"].str.split(", ")
df = df.explode("genre")
df["genre"] = df["genre"].str.strip()
```

```python
df["country"] = df["country"].str.split(", ")
df = df.explode("country")
df["country"] = df["country"].str.strip()
```

```python
df.groupby('type').size()
```

```python
df.drop(columns="date_added",inplace=True)
df.drop(columns="listed_in",inplace=True)
df.drop(columns="description",inplace=True)
```

```python
df.groupby('year_added').size() 
```

```python
df.groupby(['year_added','type']).size().unstack(fill_value=0)
```

```python
df['country'].value_counts().head(10)
```

```python
df['genre'].value_counts().head(10)
```

```python
df['rating'].value_counts()
```

```python
df.to_excel("netflix_cleaned.xlsx", index=False)
```

Key Insights:
- Netflix peaked in 2019
- Movies dominate the platform
- USA contributes highest content
- International Movies is top genre
- TV-MA is most common rating
